```python
df['title'] = df['title'].astype(str)
df['title'] = df['title'].str.replace(r'\.0$', '', regex=True)

df['title_clean'] = df['title'].str.strip().str.lower()
df['match_key'] = df['title_clean'] + '_' + df['year'].astype(str)

colliding_keys = df['match_key'][df['match_key'].duplicated(keep=False)]
runtime_str = df.loc[df['match_key'].isin(colliding_keys), 'runtime'].fillna('onbekend').astype(str)

df.loc[df['match_key'].isin(colliding_keys), 'match_key'] = (
    df.loc[df['match_key'].isin(colliding_keys), 'match_key'] + '_' + runtime_str
)

print('match_key null:', df['match_key'].isnull().sum())
print('duplicated():', df['match_key'].duplicated().sum())
```

So before I could even build a proper match_key, I noticed that some titles were not being read as text at all. Excel had turned some movie titles into actual numbers or even dates (like "1985" or "2016-11-08"), so when I tried to clean the title column with .str functions, those rows just turned into empty values instead of a proper title. I fixed this by forcing the whole title column to text first with astype(str), and then removed the extra ".0" that showed up on the ones that were read as numbers.

After that I rebuilt title_clean and match_key like before, using title and year. Some of these match_keys were still duplicate, so just like before I need to add something extra to make them unique, otherwise they'll all get matched to the wrong movie later. I used runtime again since that's the most unique column both datasets have and it almost always differs between two movies that share the same title and year.

One thing I ran into is that when runtime was missing for one of the duplicate movies, just adding it straight away actually turned the whole match_key into an empty value instead of keeping it unique, because pandas doesn't turn a missing number into the text "unknown" by itself, it stays empty even after converting to text. So I first filled the missing runtimes with the word "onbekend" before adding it to the match_key, that way even the movies with no runtime info keep a working, unique match_key instead of disappearing into empty values.
