```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("metaClean43Brightspace.xlsx")

df.head()
```

Here I read the excel file, I firstly imported openpyxl to my environment so I was able to read an excel file. Then I read the excelfile and printed out the first 5 rows.

```python
df.info()
df.isnull().sum()
```

Here I checked the info of the datasets, to see the rows and how many non-null values they have, I also checked the datatypes. I did a double check on how many null values there are in each columm. You can see that url and title got 0 null values, but summary and awards got more than half of the rows for null values. 

```python
df['rating'].unique
```

I saw that the rating columm had a problem that some of them started with '|', now we need to remove that.

```python
df['rating'] = df['rating'].str.strip(' | ').str.strip()
df['rating'].unique
```

```python
df['awards'] = df['awards'].fillna('none')
df['awards'].unique
```

For the nullvalues in the colomm awards we can assume that the movie did not win any awards. Therefore I changed the null value to none.

```python
df['studio'] = df['studio'].str.strip()
df['studio'].value_counts()
```

I still need to clean the runtime and userscore, I want to use median is that okay? Than there are some rows where director and genre is missing it is not a lot, is it possible to drop them or can we leave them open? There are studios who are written in more ways, what is the best way to clean this? 

```python
df['runtime'] = df['runtime'].fillna(df['runtime'].median())
df['userscore'] = df['userscore'].fillna(df['userscore'].median())
```

Here I filled the nullvalues of runtime and metascore with the median.

```python
df[df['genre'].isnull()]
```

Here we took a look at all the rows who had a nullvalue in genre, we did this to see if the movies are important to keep.

```python
df[df['director'].isnull()]
```

Now we did the same for directors. 

```python
df = df.dropna(subset=['director', 'genre'])
```

Here I dropped the rows that have nullvalues for director and genre.

```python
df['rating'] = df['rating'].replace({'PG--13': 'PG-13', 'PG-13`': 'PG-13'})
```

There were still some values of rating with a double -- and one with some with a '`' I changed these one so everyone of them were the same value.

```python
df['title_clean'] = df['title'].str.strip().str.lower()
df.head()
```

Here I normalised the title columm, i made sure that there were only lower cases in the titles and added that to the new colomm title_clean

```python
df['match_key'] = df['title_clean'] + '_' + df['RelDate'].dt.year.astype(str)
df['match_key'].head()
```

Here I made a match_key so that I eventually can make the new id where I can math on. This because I don't want any letters in my id (primary keys)

```python
df['match_key'].duplicated().sum()
```

I was checking for duplicates in the new colomm and it gave me three, so now im gonna check which ones they are.

```python
double_keys = df[df['match_key'].duplicated(keep=False)]
double_keys.sort_values('match_key')
```

Here I did the extra check to see if the duplicates really are the same movies. As you can see it definently are different movies, with different studios,casts and directors etc. So I'm gonna leave this duplicates, but I need to add something to them otherwise they will all get matched to the other id's. Therefore I'm gonna add the runtime, this because in the sales data runtime comes up to, this colomm is the most unique and all the runtimes differ from eachother so that is the best one to match on. 

```python
colliding_keys = df['match_key'][df['match_key'].duplicated(keep=False)]

df.loc[df['match_key'].isin(colliding_keys), 'match_key'] = (
    df.loc[df['match_key'].isin(colliding_keys), 'match_key'] 
+ '_' + df.loc[df['match_key'].isin(colliding_keys), 'runtime'].astype(str)
)    
```

```python
df['match_key'].duplicated().sum()
```

So now i added the runtime to the duplicate values and there are 0 nullvalues left.
